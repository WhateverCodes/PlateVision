```python
from pathlib import Path
import sys
root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(root))
import cv2
import matplotlib.pyplot as plt
from src.preprocessing import read_image, grayscale, enhance, rectify_plate
image = read_image(root / 'samples' / 'synthetic_scene.png')
gray = grayscale(image)
views = {'Original': cv2.cvtColor(image, cv2.COLOR_BGR2RGB), 'Grayscale': gray,
         'CLAHE': enhance(gray, clahe=True), 'Gaussian blur': cv2.GaussianBlur(gray,(3,3),0),
         'Bilateral': enhance(gray,bilateral=True), 'Sharpen': enhance(gray,sharpen=True),
         'Otsu': cv2.threshold(gray,0,255,cv2.THRESH_BINARY+cv2.THRESH_OTSU)[1],
         'Adaptive': cv2.adaptiveThreshold(gray,255,cv2.ADAPTIVE_THRESH_GAUSSIAN_C,cv2.THRESH_BINARY,31,9)}
fig, axes = plt.subplots(2,4,figsize=(16,7))
for ax,(name,view) in zip(axes.flat,views.items()):
    ax.imshow(view,cmap='gray'); ax.set_title(name); ax.axis('off')
plt.tight_layout()
```
